```python
import pandas as pd

zhvi_path = "../data/raw/zillow/Metro_zhvi_uc_sfrcondo_tier_0.33_0.67_sm_sa_month.csv"
zhvi = pd.read_csv(zhvi_path)

print("Shape:", zhvi.shape)
print("\nColumns:")
print(zhvi.columns.tolist())

zhvi.head()
```

```python
id_cols = ["RegionID", "SizeRank", "RegionName", "RegionType", "StateName"]
date_cols = [c for c in zhvi.columns if c not in id_cols]

zhvi_long = zhvi.melt(
    id_vars=id_cols,
    value_vars=date_cols,
    var_name="month",
    value_name="zhvi"
)

zhvi_long["month"] = pd.to_datetime(zhvi_long["month"])
zhvi_long["zhvi"] = pd.to_numeric(zhvi_long["zhvi"], errors="coerce")

zhvi_long = zhvi_long.rename(columns={"RegionName": "metro"})
zhvi_long = zhvi_long[["metro", "StateName", "month", "zhvi"]]

zhvi_long.head()
```

```python
zhvi_long = zhvi_long[zhvi_long["metro"] != "United States"].copy()
zhvi_long = zhvi_long.reset_index(drop=True)

zhvi_long.head()
```

```python
print("Rows:", len(zhvi_long))
print("Unique metros:", zhvi_long["metro"].nunique())
print("Date range:", zhvi_long["month"].min(), "to", zhvi_long["month"].max())
```

```python
msp_path = "../data/raw/zillow/Metro_median_sale_price_now_uc_sfrcondo_month.csv"
msp = pd.read_csv(msp_path)

print("Shape:", msp.shape)
print("\nColumns:")
print(msp.columns.tolist())

msp.head()
```

```python
id_cols = ["RegionID", "SizeRank", "RegionName", "RegionType", "StateName"]
date_cols = [c for c in msp.columns if c not in id_cols]

msp_long = msp.melt(
    id_vars=id_cols,
    value_vars=date_cols,
    var_name="month",
    value_name="median_sale_price"
)

msp_long["month"] = pd.to_datetime(msp_long["month"])
msp_long["median_sale_price"] = pd.to_numeric(msp_long["median_sale_price"], errors="coerce")

msp_long = msp_long.rename(columns={"RegionName": "metro"})
msp_long = msp_long[msp_long["metro"] != "United States"].copy()
msp_long = msp_long[["metro", "StateName", "month", "median_sale_price"]]

msp_long.head()
```

```python
print("ZHVI metros:", zhvi_long["metro"].nunique())
print("MSP metros:", msp_long["metro"].nunique())

print("ZHVI date range:", zhvi_long["month"].min(), "to", zhvi_long["month"].max())
print("MSP date range:", msp_long["month"].min(), "to", msp_long["month"].max())
```

```python
zhvi_keys = zhvi_long[["metro", "month"]].drop_duplicates()
msp_keys = msp_long[["metro", "month"]].drop_duplicates()

overlap = zhvi_keys.merge(msp_keys, on=["metro", "month"], how="inner")

print("ZHVI metro-month rows:", len(zhvi_keys))
print("MSP metro-month rows:", len(msp_keys))
print("Overlapping metro-month rows:", len(overlap))
print("Overlapping metros:", overlap["metro"].nunique())
```

```python
metro_core = zhvi_long.merge(
    msp_long,
    on=["metro", "StateName", "month"],
    how="inner"
)

print("Rows:", len(metro_core))
print("Unique metros:", metro_core["metro"].nunique())
print("Date range:", metro_core["month"].min(), "to", metro_core["month"].max())

metro_core.head()
```

```python
new_con_price_path = "../data/raw/zillow/Metro_new_con_median_sale_price_uc_sfrcondo_month.csv"
new_con_price = pd.read_csv(new_con_price_path)

print("Shape:", new_con_price.shape)
print("\nColumns:")
print(new_con_price.columns.tolist())

new_con_price.head()
```

```python
id_cols = ["RegionID", "SizeRank", "RegionName", "RegionType", "StateName"]
date_cols = [c for c in new_con_price.columns if c not in id_cols]

new_con_price_long = new_con_price.melt(
    id_vars=id_cols,
    value_vars=date_cols,
    var_name="month",
    value_name="new_con_median_sale_price"
)

new_con_price_long["month"] = pd.to_datetime(new_con_price_long["month"])
new_con_price_long["new_con_median_sale_price"] = pd.to_numeric(
    new_con_price_long["new_con_median_sale_price"], errors="coerce"
)

new_con_price_long = new_con_price_long.rename(columns={"RegionName": "metro"})
new_con_price_long = new_con_price_long[new_con_price_long["metro"] != "United States"].copy()
new_con_price_long = new_con_price_long[["metro", "StateName", "month", "new_con_median_sale_price"]]

new_con_price_long.head()
```

```python
metro_core_full = metro_core.merge(
    new_con_price_long,
    on=["metro", "StateName", "month"],
    how="inner"
)

print("Rows:", len(metro_core_full))
print("Unique metros:", metro_core_full["metro"].nunique())
print("Date range:", metro_core_full["month"].min(), "to", metro_core_full["month"].max())

metro_core_full.head()
```

```python
metro_core_full = metro_core_full.sort_values(["metro", "month"]).copy()

metro_core_full["zhvi_yoy_growth"] = (
    metro_core_full.groupby("metro")["zhvi"].pct_change(12) * 100
)

metro_core_full["median_sale_price_yoy_growth"] = (
    metro_core_full.groupby("metro")["median_sale_price"].pct_change(12) * 100
)

metro_core_full["sales_count_yoy_growth"] = (
    metro_core_full.groupby("metro")["sales_count"].pct_change(12) * 100
)

metro_core_full["new_con_sales_yoy_growth"] = (
    metro_core_full.groupby("metro")["new_con_sales_count"].pct_change(12) * 100
)

metro_core_full["new_con_price_premium"] = (
    metro_core_full["new_con_median_sale_price"] / metro_core_full["median_sale_price"]
)

metro_core_full["new_con_share_of_sales"] = (
    metro_core_full["new_con_sales_count"] / metro_core_full["sales_count"]
)

metro_core_full.head(15)
```

```python
metro_core_full[[
    "zhvi",
    "median_sale_price",
    "sales_count",
    "new_con_sales_count",
    "new_con_median_sale_price",
    "zhvi_yoy_growth",
    "median_sale_price_yoy_growth",
    "sales_count_yoy_growth",
    "new_con_sales_yoy_growth",
    "new_con_price_premium",
    "new_con_share_of_sales"
]].isna().sum()
```

```python
latest_month = metro_core_full["month"].max()
print("Latest month:", latest_month)

recent = metro_core_full[
    metro_core_full["month"] >= (latest_month - pd.DateOffset(months=12))
].copy()

metro_summary = recent.groupby(["metro", "StateName"], as_index=False).agg({
    "zhvi": "last",
    "median_sale_price": "last",
    "sales_count": "mean",
    "new_con_sales_count": "mean",
    "new_con_median_sale_price": "mean",
    "zhvi_yoy_growth": "mean",
    "median_sale_price_yoy_growth": "mean",
    "sales_count_yoy_growth": "mean",
    "new_con_sales_yoy_growth": "mean",
    "new_con_price_premium": "mean",
    "new_con_share_of_sales": "mean"
})

print("Rows:", len(metro_summary))
metro_summary.head()
```

```python
cluster_features = [
    "zhvi_yoy_growth",
    "median_sale_price_yoy_growth",
    "sales_count_yoy_growth",
    "new_con_sales_yoy_growth",
    "new_con_price_premium",
    "new_con_share_of_sales"
]

metro_cluster_data = metro_summary[["metro", "StateName"] + cluster_features].copy()
metro_cluster_data = metro_cluster_data.dropna().reset_index(drop=True)

print("Rows after dropna:", len(metro_cluster_data))
metro_cluster_data.head()
```

```python
from sklearn.preprocessing import StandardScaler

X = metro_cluster_data[[
    "zhvi_yoy_growth",
    "median_sale_price_yoy_growth",
    "sales_count_yoy_growth",
    "new_con_sales_yoy_growth",
    "new_con_price_premium",
    "new_con_share_of_sales"
]]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled shape:", X_scaled.shape)
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

results = []

for k in range(2, 7):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    results.append((k, score))

results
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
metro_cluster_data["cluster"] = kmeans.fit_predict(X_scaled)

metro_cluster_data.head()
```

```python
cluster_profile = metro_cluster_data.groupby("cluster").agg({
    "zhvi_yoy_growth": "mean",
    "median_sale_price_yoy_growth": "mean",
    "sales_count_yoy_growth": "mean",
    "new_con_sales_yoy_growth": "mean",
    "new_con_price_premium": "mean",
    "new_con_share_of_sales": "mean"
}).round(2)

cluster_profile
```

```python
metro_core_final = metro_core_full.merge(
    metro_cluster_data[["metro", "StateName", "cluster"]],
    on=["metro", "StateName"],
    how="left"
)

metro_core_final.head()
```

```python
metro_summary_final = metro_summary.merge(
    metro_cluster_data[["metro", "StateName", "cluster"]],
    on=["metro", "StateName"],
    how="left"
)

metro_summary_final.head()
```

```python
metro_core_final.to_csv("../data/processed/metro_core_final.csv", index=False)
metro_summary_final.to_csv("../data/processed/metro_summary_final.csv", index=False)

print("Files saved successfully.")
```
